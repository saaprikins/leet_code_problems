- Let $n$ be the number of steps in the staircase, $n_{i}$ the number of steps per $i_{th}$ climb, and $k$ the number of ways to climb the staircase
- For $n=1$, the solution is trivial, $k=1$
- For $n=2$, the solution is either climb one plus one steps, or two straight, leaving $k=2$ in total
- Let $n=3$. One might either take one step, and have two steps remaining. Which might be covered in $k=2$ ways. Alternatively, one might take two steps first, and have one step remaining. Which might be covered in $k=1$ way. In total, this sums to $k=3$ ways.
- Let $n=N$. One might either take one step, and have $N-1$ steps remaining. Which might be covered in $k(N-1)$ ways. Alternatively, one might take two steps first, and have $N-2$ steps remaining. Which might be covered in $k(N-2)$ ways. In total, this sums to $k(N-1)$ + $k(N-2)$ ways.
- Hence, we are dealing with Fibonacci figures, starting from 3rd figure (1)

```python
class Solution:
    def climbStairs(self, n: int) -> int:

        # space O(1)
        # fix the first three figures
        prev_prev, prev, curr = 0, 1, 1

        for i in range(2, n + 2):
            curr = prev_prev + prev
            prev_prev = prev
            prev = curr

        return curr
```
