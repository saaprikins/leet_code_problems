```python
import numpy as np
from typing import List
```

```python
class Solution:
    def __init__(self, coins: List[List[int]] = None):

        # a few check on validity of entries
        if coins is None:
            raise ValueError("empty coins list!")
        if any(len(row) == 0 for row in coins):
            raise ValueError("missing values are present!")
        if not (1 <= len(coins) <= 500):
            raise ValueError("invalid shape (rows)!")
        if not all(1 <= len(row) <= 500 for row in coins):
            raise ValueError("invalid shape (columns)!")
```

```python
coins1 = [[12, 3, 4], [5, 6, 9], []]
coins2 = [[]]
# print(np.matrix(coins2))
```

```python
sol = Solution(coins=coins1)
```
